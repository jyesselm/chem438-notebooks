# Python Variable Assignment Problems

Problem 1
Create a 1D NumPy array of integers from 0 to 9. Print the array.


Problem 2
Create a 3x3 array of random floats between 0 and 1 using NumPy. Print the array.


Problem 3
Create a 1D array [10, 20, 30, 40, 50]. Compute its mean, median, and standard deviation using NumPy.


Problem 4
Generate an array of 100 random numbers from a normal distribution with mean 0 and std 1. Compute the mean and standard deviation of the generated array.


Problem 5
Create two 3x3 arrays and compute their dot product using NumPy.


Problem 6
Create an array x from 0 to 10 using linspace with 100 points. Plot y = x^2.


Problem 7
Create an array x from -5 to 5 using arange with 0.5 step. Plot y = 2x + 1.


Problem 8
Create an array x from 0 to 2π using linspace with 200 points. Plot y = cos(x) and y = sin(x) on the same graph.


Problem 9
Create an array x from 0 to 5 using linspace with 100 points. Plot y = e^(-x).


Problem 10
Create an array x from -10 to 10 using arange with 0.1 step. Plot y = 1 / (1 + e^(-x)).


Problem 11
Create an array x from -5 to 5 using linspace with 100 points. Plot y = |x| (absolute value function).


Problem 12
Use SciPy’s optimize module to find the minimum of the function f(x) = (x - 3)**2 + 2.


Problem 13
Plot the quadratic function f(x)= (x-2)^2 + 1 over [-2,6] with 400 points. Find its minimum with SciPy and mark the minimum on the plot.


Problem 14
Consider f(x) = sin(x) + 0.2x on [0, 6π] with 1000 points. Find the x that minimizes f using SciPy, then plot the function and mark the minimum.


Problem 15
Generate noisy exponential decay data y = A*exp(-kx) with A=5, k=0.7 from 0 to 5 with 60 points and noise. Fit A and k with curve_fit and plot.


Problem 16
Define f(x) = x^4 - 3x^3 + 2 and search for its global minimum on [-1, 4] with SciPy. Plot f and mark the minimum.


Problem 17
Generate noisy logistic data y = L/(1+exp(-k(x-x0))) + ε with L=10,k=1.2,x0=5, from 0 to 10 with 120 points and noise. Fit L,k,x0 and plot.


Problem 18
Generate 1000 samples from a normal distribution (mean=50, std=10) and perform a one-sample t-test, to check if the mean differs significantly from 50.
