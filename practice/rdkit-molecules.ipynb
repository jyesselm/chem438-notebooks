# Python Variable Assignment Problems

Problem 1
Create an RDKit molecule object for aspirin (acetylsalicylic acid) using its SMILES string. Generate a 2D depiction of the molecule and display it.


Problem 2
Calculate the molecular weight and LogP of caffeine. Use the SMILES string "CN1C=NC2=C1C(=O)N(C(=O)N2C)C" for caffeine. Use descriptors from the rdkit.Chem.Descriptors module.


Problem 3
Create a 2D depiction of nicotine and add explicit hydrogens before drawing. Use SMILES "CN1CCCC1C2=CN=CC=C2" for nicotine. Display both the H-implicit and H-explicit images side-by-side.


Problem 4<br>Count the number of hydrogen bond donors and acceptors in paracetamol (acetaminophen). Use the SMILES string "CC(=O)NC1=CC=C(C=C1)O" for paracetamol. Use descriptors from the rdkit.Chem.Descriptors<br>module.


Problem 5
Count aromatic atoms and aromatic bonds in naproxen. Use SMILES "CC(C)c1ccc2cc(ccc2c1)C(C)C(=O)O" for naproxen.


Problem 6<br>Create a grid of 2D depictions for four common over-the-counter drugs: aspirin, ibuprofen, acetaminophen (paracetamol), and naproxen. Use their SMILES strings to create RDKit molecule objects, then<br>use RDKit's Draw.MolsToGridImage function to create a 2x2 grid of these molecules. Label each molecule with its name.


Problem 7
Compute Gasteiger partial charges after adding explicit hydrogens for acetaminophen and print the charge on O and N atoms. Use SMILES "CC(=O)NC1=CC=C(C=C1)O" for acetaminophen.


Problem 8
Count ring sizes present in morphine by listing unique ring sizes found in the SSSR. Use SMILES "CN1CCC23C4C1CC5=CC(=O)OC(C=C5C2=CC=C3O)O" for morphine.


Problem 9
Calculate the number of carbon and oxygen atoms in the aspirin molecule. Use a loop to iterate over the atoms in the molecule and count the occurrences of carbon and oxygen.


Problem 10
Find all atoms in the caffeine molecule that are part of a ring system. Use a loop to iterate over the atoms and check if they are in a ring. Print the atomic number and symbol for each ring atom.


Problem 11
Highlight specific atoms in a molecule using RDKit. Create a molecule of benzene, then highlight three alternating carbon atoms in red.


Problem 12
Set custom atom labels for a molecule using RDKit. Create a molecule of ethanol and label the oxygen atom as "OH" and the carbon atoms as "C1" and "C2".


Problem 13
Create a molecule of benzene and label every other carbon atom with its atomic number. Display the resulting molecule.


Problem 14
Perform molecular fragmentation using RDKit. Create a molecule of aspirin, break the ester bond, and display the resulting fragments.
