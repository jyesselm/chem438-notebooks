# Solutions — Variables & arithmetic

```
# Problem 1
# Assign the values 7 and 3 to variables `a` and `b`, respectively. Compute their sum and assign it to a variable `sum_ab`.
a = 7
b = 3
sum_ab = a + b
sum_ab  # Output: 10
```
