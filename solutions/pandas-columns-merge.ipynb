Problem 1
Create a DataFrame with 'First_Name' and 'Last_Name' columns. Add a new column 'Full_Name' that combines the first and last names.


```
data = {
    "First_Name": ["John", "Jane", "Mike", "Emily"],
    "Last_Name": ["Doe", "Smith", "Johnson", "Brown"],
}

# Solution
df = pd.DataFrame(data)
df["Full_Name"] = df["First_Name"] + " " + df["Last_Name"]
print(df)
```

Problem 2
Create a DataFrame with 'Name' and 'Salary' columns. Add a new column 'Salary_Category' that categorizes salaries as 'Low' if < 50000, 'Medium' if between 50000 and 100000, and 'High' if > 100000.


```
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Salary": [45000, 75000, 110000, 60000],
}

# Solution
df = pd.DataFrame(data)
df["Salary_Category"] = ""
for index, row in df.iterrows():
    if row["Salary"] < 50000:
        df.at[index, "Salary_Category"] = "Low"
    elif 50000 <= row["Salary"] <= 100000:
        df.at[index, "Salary_Category"] = "Medium"
    else:
        df.at[index, "Salary_Category"] = "High"
print(df)
```

Problem 6

```
data = {
    "Text": [
        "Hello world",
        "Data science is fun",
        "Python pandas",
        "New column creation",
    ]
}

# Solution
df = pd.DataFrame(data)
df["Word_Count"] = 0
for index, row in df.iterrows():
    df.at[index, "Word_Count"] = len(row["Text"].split())
print(df)
```

Problem 10
Create a DataFrame with 'Product' and 'Price' columns. Use iterrows to create a new column 'Discount' that's 10% off the original price for items over $50, and 5% off for items $50 or under.


```
data = {"Product": ["A", "B", "C", "D", "E"], "Price": [40, 60, 30, 80, 50]}

# Solution
df = pd.DataFrame(data)
df["Discount"] = 0
for index, row in df.iterrows():
    if row["Price"] > 50:
        df.at[index, "Discount"] = row["Price"] * 0.1
    else:
        df.at[index, "Discount"] = row["Price"] * 0.05
print(df)
```

Problem 15
Create a DataFrame with 'Item' and 'Category' columns. Use iterrows to create a dictionary where keys are categories and values are lists of items in that category.


```
data = {
    "Item": ["Apple", "Carrot", "Banana", "Broccoli", "Cherry"],
    "Category": ["Fruit", "Vegetable", "Fruit", "Vegetable", "Fruit"],
}

# Solution
df = pd.DataFrame(data)
category_dict = {}
for index, row in df.iterrows():
    if row["Category"] not in category_dict:
        category_dict[row["Category"]] = []
    category_dict[row["Category"]].append(row["Item"])
print(category_dict)
```

Problem 16
Create a DataFrame with 'Name' and 'Grades' columns (Grades as a string of comma-separated values). Use iterrows to calculate and add an 'Average' column with the average grade for each student.


```
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Grades": ["85,90,78", "92,88,95", "75,80,82"],
}

# Solution
df = pd.DataFrame(data)
df["Average"] = 0
for index, row in df.iterrows():
    grades = [int(grade) for grade in row["Grades"].split(",")]
    df.at[index, "Average"] = sum(grades) / len(grades)
print(df)
```

Problem 17
Create a DataFrame with 'City' and 'Temperature' columns (in Celsius). Use iterrows to create a new 'Fahrenheit' column that converts the temperature to Fahrenheit.


```
data = {
    "City": ["New York", "London", "Tokyo", "Sydney", "Moscow"],
    "Temperature": [25, 18, 30, 22, 15],
}

# Solution
df = pd.DataFrame(data)
df["Fahrenheit"] = 0
for index, row in df.iterrows():
    df.at[index, "Fahrenheit"] = (row["Temperature"] * 9 / 5) + 32
print(df)
```

Problem 18
Create two DataFrames: one with 'ID' and 'Name', and another with 'ID' and 'Score'. Merge them on 'ID' to create a single DataFrame with all columns. Print the result.


```
data1 = {
    "ID": [1, 2, 3, 4],
    "Name": ["Alice", "Bob", "Charlie", "David"],
}
data2 = {
    "ID": [2, 4, 3, 1],
    "Score": [85, 92, 78, 88],
}

# Solution
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
merged_df = pd.merge(df1, df2, on="ID")
print(merged_df)
```

Problem 19<br>Create two DataFrames: one with 'EmployeeID' and 'Department', and another with 'EmployeeID' and 'Salary'. Perform a left merge to include all employees and their salaries (if available). Print the<br>result.


```
data_dept = {
    "EmployeeID": [101, 102, 103, 104],
    "Department": ["HR", "IT", "Sales", "Marketing"],
}
data_salary = {
    "EmployeeID": [102, 104, 105],
    "Salary": [60000, 65000, 70000],
}

# Solution
df_dept = pd.DataFrame(data_dept)
df_salary = pd.DataFrame(data_salary)
left_merged = pd.merge(df_dept, df_salary, on="EmployeeID", how="left")
print(left_merged)
```

Problem 20<br>Create two DataFrames: one with 'StudentID' and 'Name', and another with 'StudentID' and 'Club'. Perform an outer merge to include all students and all clubs, matching where possible. Print the<br>result.


```
data_students = {
    "StudentID": [1, 2, 3, 4],
    "Name": ["Alice", "Bob", "Charlie", "David"],
}
data_clubs = {
    "StudentID": [3, 4, 5, 6],
    "Club": ["Chess", "Robotics", "Drama", "Music"],
}

# Solution
df_students = pd.DataFrame(data_students)
df_clubs = pd.DataFrame(data_clubs)
outer_merged = pd.merge(df_students, df_clubs, on="StudentID", how="outer")
print(outer_merged)
```
