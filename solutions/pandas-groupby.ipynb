Problem 1
Create a DataFrame with 'Name', 'Age', and 'City' columns. Filter to show only people who are either over 30 or live in New York. print the final dataframe


```
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Age": [25, 35, 28, 40, 22],
    "City": ["New York", "Los Angeles", "Chicago", "New York", "Boston"],
}

# Solution
df = pd.DataFrame(data)
filtered_df = df[(df["Age"] > 30) | (df["City"] == "New York")]
print(filtered_df)
```

Problem 2<br>Create a DataFrame with 'Product', 'Category', 'Price', and 'Stock' columns. Filter to show products that are either Electronics costing more than $500 or have less than 10 items in stock. print the<br>final dataframe


```
data = {
    "Product": ["Laptop", "T-shirt", "Smartphone", "Jeans", "Tablet"],
    "Category": ["Electronics", "Clothing", "Electronics", "Clothing", "Electronics"],
    "Price": [800, 25, 600, 50, 300],
    "Stock": [15, 100, 20, 80, 5],
}

# Solution
df = pd.DataFrame(data)
filtered_df = df[
    ((df["Category"] == "Electronics") & (df["Price"] > 500)) | (df["Stock"] < 10)
]
print(filtered_df)
```

Problem 6<br>Create a DataFrame with 'Department', 'Employee', 'Salary', and 'Years_of_Service' columns. Use groupby and agg to calculate the average salary and total years of service for each department. print<br>the final dataframe


```
data = {
    "Department": ["HR", "IT", "HR", "IT", "Finance", "Finance"],
    "Employee": ["Alice", "Bob", "Charlie", "David", "Eve", "Frank"],
    "Salary": [50000, 60000, 52000, 65000, 55000, 58000],
    "Years_of_Service": [3, 5, 2, 4, 3, 6],
}

# Solution
df = pd.DataFrame(data)
result = (
    df.groupby("Department")
    .agg({"Salary": "mean", "Years_of_Service": "sum"})
    .rename(
        columns={"Salary": "Avg_Salary", "Years_of_Service": "Total_Years_of_Service"}
    )
)
print(result)
```

Problem 7<br>Create a DataFrame with 'Product', 'Category', 'Sales', and 'Profit' columns. Use groupby and agg to find the total sales, average profit, and number of products for each category. print the final<br>dataframe


```
data = {
    "Product": ["A", "B", "C", "D", "E", "F"],
    "Category": [
        "Electronics",
        "Clothing",
        "Electronics",
        "Clothing",
        "Books",
        "Books",
    ],
    "Sales": [1000, 1200, 800, 1500, 500, 600],
    "Profit": [200, 150, 180, 250, 100, 120],
}

# Solution
df = pd.DataFrame(data)
result = (
    df.groupby("Category")
    .agg({"Sales": "sum", "Profit": "mean", "Product": "count"})
    .rename(
        columns={
            "Sales": "Total_Sales",
            "Profit": "Avg_Profit",
            "Product": "Number_of_Products",
        }
    )
)
print(result)
```

Problem 8
Create a DataFrame with 'Class', 'Student', and 'Score', Use groupby and agg to compute the average score and number of students in each class. print the final dataframe


```
data = {
    "Class": ["A", "A", "B", "B", "B", "C", "C"],
    "Student": ["Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Gina"],
    "Score": [85, 92, 78, 88, 91, 73, 80],
}

# Solution
df = pd.DataFrame(data)
result = df.groupby("Class").agg(
    Avg_Score=("Score", "mean"),
    Num_Students=("Student", "count")
)
print(result)
```

Problem 9
Create a DataFrame with 'Region', 'OrderID', 'Sales', and 'Discount'. Use groupby and agg to find total sales, average discount, and number of orders per region. print the final dataframe


```
data = {
    "Region": ["East", "West", "East", "South", "West", "South", "East"],
    "OrderID": [101, 102, 103, 104, 105, 106, 107],
    "Sales": [200, 150, 350, 400, 180, 220, 300],
    "Discount": [0.1, 0.0, 0.15, 0.05, 0.2, 0.1, 0.0],
}

# Solution
df = pd.DataFrame(data)
result = df.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Avg_Discount=("Discount", "mean"),
    Num_Orders=("OrderID", "count")
)
print(result)
```

Problem 10<br>Create a DataFrame with 'User', 'Session', 'Minutes', and 'Page', Use groupby and agg to compute total minutes, average minutes per session, and number of unique (hint use nunique) pages per user.<br>print the final dataframe


```
data = {
    "User": ["u1", "u1", "u2", "u2", "u2", "u3"],
    "Session": [1, 2, 1, 1, 2, 1],
    "Minutes": [30, 45, 10, 25, 40, 50],
    "Page": ["Home", "Products", "Home", "Cart", "Checkout", "Home"],
}

# Solution
df = pd.DataFrame(data)
result = df.groupby("User").agg(
    Total_Minutes=("Minutes", "sum"),
    Avg_Minutes=("Minutes", "mean"),
    Unique_Pages=("Page", "nunique")
)
print(result)
```

Problem 11<br>Create a DataFrame with 'Category', 'Brand', 'Price', and 'Stock', First add a column 'Inventory_Value' = Price * Stock, then groupby Category and Brand and agg total stock and total inventory value.<br>print the final dataframe


```
data = {
    "Category": ["Electronics", "Electronics", "Clothing", "Clothing", "Clothing"],
    "Brand": ["A", "B", "C", "C", "D"],
    "Price": [800, 600, 40, 50, 70],
    "Stock": [5, 8, 100, 60, 40],
}

# Solution
df = pd.DataFrame(data)
df["Inventory_Value"] = df["Price"] * df["Stock"]
result = df.groupby(["Category", "Brand"]).agg(
    Total_Stock=("Stock", "sum"),
    Total_Value=("Inventory_Value", "sum")
)
print(result)
```

Problem 12
Create a DataFrame with 'Genre', 'Title', and 'Rating', Use groupby and agg to compute average rating, median rating, and number of titles for each genre. print the final dataframe


```
data = {
    "Genre": ["Sci-Fi", "Drama", "Drama", "Comedy", "Sci-Fi", "Comedy"],
    "Title": ["T1", "T2", "T3", "T4", "T5", "T6"],
    "Rating": [4.5, 3.8, 4.2, 3.5, 4.9, 3.9],
}

# Solution
df = pd.DataFrame(data)
result = df.groupby("Genre").agg(
    Avg_Rating=("Rating", "mean"),
    Median_Rating=("Rating", "median"),
    Num_Titles=("Title", "count")
)
print(result)
```

Problem 15<br>Create a DataFrame with 'Product', 'Category', 'Price', and 'Quantity', Add a new column 'Revenue' = Price * Quantity. Filter to only include rows where Revenue > 500, then group by Category to<br>compute total revenue and average price. print the final dataframe


```
data = {
    "Product": ["A", "B", "C", "D", "E"],
    "Category": ["Electronics", "Electronics", "Clothing", "Clothing", "Electronics"],
    "Price": [200, 150, 40, 80, 300],
    "Quantity": [3, 5, 20, 10, 1],
}

# Solution
df = pd.DataFrame(data)
df["Revenue"] = df["Price"] * df["Quantity"]
filtered_df = df[df["Revenue"] > 500]
result = filtered_df.groupby("Category").agg(
    Total_Revenue=("Revenue", "sum"),
    Avg_Price=("Price", "mean")
)
print(result)
```

Problem 16<br>Create a DataFrame with 'Date', 'Store', 'Product', 'Sales', and 'Returns', Add a 'Net_Sales' column = Sales - Returns. Filter to only include rows with Net_Sales >= 100, then group by Store and<br>Product to compute total net sales and total number of transactions. print the final dataframe


```
data = {
    "Date": pd.date_range("2023-01-01", periods=6),
    "Store": ["S1", "S1", "S2", "S2", "S3", "S3"],
    "Product": ["P1", "P2", "P1", "P3", "P2", "P4"],
    "Sales": [150, 80, 120, 200, 90, 50],
    "Returns": [20, 10, 15, 30, 5, 0],
}

# Solution
df = pd.DataFrame(data)
df["Net_Sales"] = df["Sales"] - df["Returns"]
filtered_df = df[df["Net_Sales"] >= 100]
result = filtered_df.groupby(["Store", "Product"]).agg(
    Total_Net_Sales=("Net_Sales", "sum"),
    Num_Transactions=("Date", "count")
)
print(result)
```

Problem 17<br>Create a DataFrame with 'Employee', 'Department', 'Monthly_Salary', and 'Months_Worked', Add an 'Annual_Salary' column = Monthly_Salary * Months_Worked, Filter to only include employees with<br>Annual_Salary > 40000, then group by Department to compute total and average annual salary. print the final dataframe


```
data = {
    "Employee": ["A", "B", "C", "D", "E"],
    "Department": ["IT", "HR", "IT", "Finance", "Finance"],
    "Monthly_Salary": [3500, 3000, 4000, 4500, 2000],
    "Months_Worked": [12, 10, 12, 9, 12],
}

# Solution
df = pd.DataFrame(data)
df["Annual_Salary"] = df["Monthly_Salary"] * df["Months_Worked"]
filtered_df = df[df["Annual_Salary"] > 40000]
result = filtered_df.groupby("Department").agg(
    Total_Annual_Salary=("Annual_Salary", "sum"),
    Avg_Annual_Salary=("Annual_Salary", "mean")
)
print(result)
```
