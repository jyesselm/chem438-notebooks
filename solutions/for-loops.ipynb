Problem 1
Create a list `fruits` with elements "apple", "banana", "cherry", "date". Use a for loop with enumerate() to print each fruit with its index.


```
# Solution
fruits = ["apple", "banana", "cherry", "date"]
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")
```

Problem 2
Create a list `numbers` with elements 1 to 10. Use a for loop to save all the even numbers in a new list `even_numbers` and print it.


```
# Solution
numbers = list(range(1, 11))
even_numbers = []
for num in numbers:
    if num % 2 == 0:
        even_numbers.append(num)
print(even_numbers)
```

Problem 22<br>Create a list `words` with elements "hello", "world", "python", "code", "loop". Use a for loop to count vowels and consonants and save it in a variable `vowel_count` and `consonant_count` and print<br>it.


```
# Solution
words = ["hello", "world", "python", "code", "loop"]
vowel_count = 0
consonant_count = 0
vowels = "aeiou"
for word in words:
    for letter in word:
        if letter.lower() in vowels:
            vowel_count += 1
        else:
            consonant_count += 1
print(f"Total vowels: {vowel_count}, Total consonants: {consonant_count}")
```

Problem 23
Create a list `scores` with elements 85, 92, 78, 88, 95, 82. Use a for loop to track both the highest and lowest scores and save it in a variable `highest` and `lowest` and print it.


```
# Solution
scores = [85, 92, 78, 88, 95, 82]
highest = scores[0]
lowest = scores[0]
for score in scores:
    if score > highest:
        highest = score
    if score < lowest:
        lowest = score
    print(f"Score {score}: highest = {highest}, lowest = {lowest}")
```

Problem 27<br>Create a list `measurements` with elements 5.2, 3.8, 7.1, 4.5, 6.9, 2.3. Use a for loop to find the range (difference between highest and lowest values) of all measurements seen so far and print it<br>after each measurement.


```
# Solution
measurements = [5.2, 3.8, 7.1, 4.5, 6.9, 2.3]
highest = measurements[0]
lowest = measurements[0]
for measurement in measurements:
    if measurement > highest:
        highest = measurement
    if measurement < lowest:
        lowest = measurement
    range_value = highest - lowest
    print(f"Measurement {measurement}: range so far = {range_value:.1f}")
```

Problem 41
Create a dictionary `temperatures` with days of the week as keys and temperatures as values, come up with your own data. Use a for loop to find and print the hottest day.


```
# Solution
temperatures = {"Mon": 70, "Tue": 75, "Wed": 80, "Thu": 72, "Fri": 78}
hottest_day = ""
max_temp = -999
for day, temp in temperatures.items():
    if temp > max_temp:
        max_temp = temp
        hottest_day = day
print(f"Hottest day: {hottest_day} ({max_temp}°F)")
```
