Problem 28<br>Create two lists `list1` and `list2` with elements [1, 2, 3] and [4, 5, 6] respectively. Use a for loop to create a new list `sum_lists` containing the sum of corresponding elements. Print all three<br>lists.

Problem 32
Create a list `numbers` with elements 1 to 10. Use a for loop to create two new lists: `odd_numbers` and `even_numbers`. Print all three lists.


```
# Solution
numbers = list(range(1, 11))
odd_numbers = []
even_numbers = []
for num in numbers:
    if num % 2 == 0:
        even_numbers.append(num)
    else:
        odd_numbers.append(num)
print(numbers)
print(odd_numbers)
print(even_numbers)
```

Problem 41
Create a list `temperatures` with elements 20, 25, 30, 35, 40. Use a for loop to convert each temperature to Fahrenheit and print the result, equation: fahrenheit = (temp * 9 / 5) + 32
