Problem 1
Load the tips dataset from seaborn. Compare tips vs total bill by smoker status (scatter).


```
# Solution
tips = sns.load_dataset("tips")
sns.relplot(data=tips, x="total_bill", y="tip", hue="smoker", kind="scatter")
```

Problem 2
Load the tips dataset from seaborn. Compare tips vs total bill by time (marker style) (scatter).


```
# Solution
tips = sns.load_dataset("tips")
sns.relplot(
    data=tips, x="total_bill", y="tip", hue="time", style="time", kind="scatter"
)
```

Problem 3
Load the tips dataset from seaborn. Compare tip percentage across day and smoker (point plot).


```
# Solution
tips = sns.load_dataset("tips").assign(
    tip_pct=lambda d: 100 * d["tip"] / d["total_bill"]
)
sns.catplot(data=tips, x="day", y="tip_pct", hue="smoker", kind="point", dodge=True)
```

Problem 4
Load the tips dataset from seaborn. Compare total bill by day and sex (box).


```
# Solution
tips = sns.load_dataset("tips")
sns.catplot(data=tips, x="day", y="total_bill", hue="sex", kind="box")
```

Problem 5
Load the tips dataset from seaborn. Compare total bill by day and smoker (violin).


```
# Solution
tips = sns.load_dataset("tips")
sns.catplot(
    data=tips, x="day", y="total_bill", hue="smoker", kind="violin", split=False
)
```

Problem 6
Load the penguins dataset from seaborn. Compare distribution of flipper lengths by species (hist).


```
# Solution
penguins = sns.load_dataset("penguins").dropna()
sns.displot(penguins, x="flipper_length_mm", hue="species", bins=20)
```

Problem 15
Load the healthexp dataset from seaborn. Compare Spending vs Life Expectancy for US and Germany (line).


```
# Solution
healthexp = sns.load_dataset("healthexp").sort_values("Year")
subset = healthexp[healthexp["Country"].isin(["United States", "Germany"])]
sns.relplot(
    data=subset, x="Spending_USD", y="Life_Expectancy", hue="Country", kind="line"
)
```
