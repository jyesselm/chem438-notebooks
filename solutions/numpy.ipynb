Problem 13
Plot the quadratic function f(x)= (x-2)^2 + 1 over [-2,6] with 400 points. Find its minimum with SciPy and mark the minimum on the plot.


```
# Solution
def f_quadratic(x):
    return (x - 2) ** 2 + 1


xs = np.linspace(-2, 6, 400)
ys = f_quadratic(xs)
res = optimize.minimize_scalar(f_quadratic, bounds=(-2, 6), method="bounded")
xmin, ymin = res.x, res.fun
plt.figure()
plt.plot(xs, ys, label="f(x)=(x-2)^2+1")
plt.scatter([xmin], [ymin], color="red", zorder=3, label=f"min at x≈{xmin:.2f}")
plt.title("Quadratic Function and Minimum")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
```

Problem 14
Consider f(x) = sin(x) + 0.2x on [0, 6π] with 1000 points. Find the x that minimizes f using SciPy, then plot the function and mark the minimum.


```
# Solution
def f_sin_plus_linear(x):
    return np.sin(x) + 0.2 * x


xs = np.linspace(0, 6 * np.pi, 1000)
ys = f_sin_plus_linear(xs)
res = optimize.minimize_scalar(
    f_sin_plus_linear, bounds=(0, 6 * np.pi), method="bounded"
)
plt.figure()
plt.plot(xs, ys, label="f(x)=sin(x)+0.2x")
plt.scatter(
    [res.x],
    [f_sin_plus_linear(res.x)],
    color="red",
    zorder=3,
    label=f"min at x≈{res.x:.2f}",
)
plt.title("Minimum of sin(x)+0.2x")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
```

Problem 15
Generate noisy exponential decay data y = A*exp(-kx) with A=5, k=0.7 from 0 to 5 with 60 points and noise. Fit A and k with curve_fit and plot.


```
# Solution
def exp_decay(x, A, k):
    return A * np.exp(-k * x)


x = np.linspace(0, 5, 60)
y_true = exp_decay(x, 5.0, 0.7)
y_noisy = y_true + np.random.normal(0, 0.2, size=x.size)
popt, pcov = curve_fit(exp_decay, x, y_noisy, p0=(4.0, 0.5))
y_hat = exp_decay(x, *popt)
plt.figure()
plt.scatter(x, y_noisy, s=20, label="noisy data")
plt.plot(x, y_true, linestyle="--", label="true")
plt.plot(x, y_hat, label=f"fit: A={popt[0]:.2f}, k={popt[1]:.2f}")
plt.title("Exponential Decay Fit")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
```

Problem 16
Define f(x) = x^4 - 3x^3 + 2 and search for its global minimum on [-1, 4] with SciPy. Plot f and mark the minimum.


```
# Solution
def f_poly(x):
    return x**4 - 3 * x**3 + 2


xs = np.linspace(-1, 4, 600)
ys = f_poly(xs)
res = optimize.minimize_scalar(f_poly, bounds=(-1, 4), method="bounded")
plt.figure()
plt.plot(xs, ys, label="f(x)=x^4 - 3x^3 + 2")
plt.scatter(
    [res.x], [f_poly(res.x)], color="red", zorder=3, label=f"min at x≈{res.x:.3f}"
)
plt.title("Polynomial Minimum (bounded search)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
```

Problem 17
Generate noisy logistic data y = L/(1+exp(-k(x-x0))) + ε with L=10,k=1.2,x0=5, from 0 to 10 with 120 points and noise. Fit L,k,x0 and plot.


```
# Solution
def logistic(x, L, k, x0):
    return L / (1 + np.exp(-k * (x - x0)))


x = np.linspace(0, 10, 120)
y_true = logistic(x, 10.0, 1.2, 5.0)
y_noisy = y_true + np.random.normal(0, 0.3, size=x.size)
popt, _ = curve_fit(logistic, x, y_noisy, p0=(8.0, 1.0, 4.0), maxfev=10000)
y_hat = logistic(x, *popt)
plt.figure()
plt.scatter(x, y_noisy, s=15, label="noisy data")
plt.plot(x, y_true, linestyle="--", label="true")
plt.plot(x, y_hat, label=f"fit: L={popt[0]:.2f}, k={popt[1]:.2f}, x0={popt[2]:.2f}")
plt.title("Logistic Curve Fitting")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
```
