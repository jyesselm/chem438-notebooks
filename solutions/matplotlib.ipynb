Problem 1
Create a DataFrame with 'Month' and 'Sales' columns. Plot a line graph of sales over the months. make sure to include a title, x and y labels, and a legend.


```
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [1000, 1200, 900, 1500, 1800, 1600],
}

# Solution
df = pd.DataFrame(data)
fig, ax = plt.subplots()
ax.plot(df["Month"], df["Sales"], marker="o", label="Sales")
ax.set_title("Monthly Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Sales ($)")
ax.legend()
```

Problem 13
Create a scatter plot of student heights vs weights, color the points based on gender (blue for male, red for female), add a title "Student Height vs Weight", label axes, and include a legend


```
data = {
    "Height": [165, 170, 155, 172, 180, 160, 175, 158, 182, 168],
    "Weight": [60, 65, 55, 70, 80, 58, 72, 56, 85, 63],
    "Gender": ["F", "M", "F", "M", "M", "F", "M", "F", "M", "F"],
}

# Solution
df = pd.DataFrame(data)
fig, ax = plt.subplots(figsize=(10, 6))
colors = {"M": "blue", "F": "red"}
for gender in colors:
    subset = df[df["Gender"] == gender]
    ax.scatter(subset["Height"], subset["Weight"], c=colors[gender], label=gender)
ax.set_title("Student Height vs Weight")
ax.set_xlabel("Height (cm)")
ax.set_ylabel("Weight (kg)")
ax.legend()
```

Problem 14
Create a DataFrame with 'Month', 'Temperature', and 'Rainfall' columns. Create two subplots: a line plot of temperature (left) and a bar plot of rainfall (right). include titles and axis labels


```
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Temperature": [2, 4, 8, 13, 18, 22],
    "Rainfall": [78, 65, 70, 55, 48, 40],
}

# Solution
df = pd.DataFrame(data)
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(df["Month"], df["Temperature"], marker="o")
axes[0].set_title("Monthly Temperature")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Temperature (°C)")
axes[1].bar(df["Month"], df["Rainfall"])
axes[1].set_title("Monthly Rainfall")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Rainfall (mm)")
```

Problem 15<br>Create a DataFrame with 'Score' values. Make two subplots: a histogram of raw scores (left) and a bar plot of counts per grade bucket (right) with buckets F(<60), D(60–69), C(70–79), B(80–89),<br>A(90–100). include titles and axis labels


```
data = {
    "Score": [55, 62, 77, 83, 91, 68, 74, 88, 95, 100, 59, 72, 81, 86, 93, 65, 70, 79, 84, 90],
}

# Solution
df = pd.DataFrame(data)
bins = [0, 60, 70, 80, 90, 101]
labels = ["F", "D", "C", "B", "A"]
df["Grade"] = pd.cut(df["Score"], bins=bins, labels=labels, right=False)
counts = df["Grade"].value_counts().reindex(labels, fill_value=0)
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].hist(df["Score"], bins=10, edgecolor="black")
axes[0].set_title("Score Distribution")
axes[0].set_xlabel("Score")
axes[0].set_ylabel("Count")
axes[1].bar(counts.index.astype(str), counts.values)
axes[1].set_title("Counts per Grade")
axes[1].set_xlabel("Grade")
axes[1].set_ylabel("Count")
```
