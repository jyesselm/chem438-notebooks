Problem 1<br>Create a DataFrame with 'OrderID', 'Region', 'Category', 'Price', and 'Quantity', Add a new column 'Revenue' = Price * Quantity. Filter to include only orders where Quantity >= 3, then group by<br>Region and Category to compute total revenue and total quantity. print the final dataframe


```
import pandas as pd

data = {
    "OrderID": [1, 2, 3, 4, 5, 6],
    "Region": ["East", "East", "West", "West", "South", "South"],
    "Category": ["Electronics", "Furniture", "Electronics", "Furniture", "Electronics", "Furniture"],
    "Price": [200, 150, 300, 80, 120, 60],
    "Quantity": [2, 4, 5, 1, 3, 6],
}

# Solution
df = pd.DataFrame(data)
df["Revenue"] = df["Price"] * df["Quantity"]
filtered_df = df[df["Quantity"] >= 3]
result = filtered_df.groupby(["Region", "Category"]).agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Quantity=("Quantity", "sum")
)
print(result)
```

Problem 2<br>Create a DataFrame with 'Employee', 'Department', 'Hourly_Rate', and 'Hours_Worked', Add a 'Weekly_Pay' column = Hourly_Rate * Hours_Worked, Filter to only include employees with Weekly_Pay >= 1000,<br>then group by Department to compute average weekly pay and total hours worked. print the final dataframe


```
import pandas as pd

data = {
    "Employee": ["A", "B", "C", "D", "E"],
    "Department": ["IT", "IT", "HR", "Finance", "Finance"],
    "Hourly_Rate": [25, 40, 20, 30, 15],
    "Hours_Worked": [40, 35, 50, 45, 70],
}

# Solution
df = pd.DataFrame(data)
df["Weekly_Pay"] = df["Hourly_Rate"] * df["Hours_Worked"]
filtered_df = df[df["Weekly_Pay"] >= 1000]
result = filtered_df.groupby("Department").agg(
    Avg_Weekly_Pay=("Weekly_Pay", "mean"),
    Total_Hours=("Hours_Worked", "sum")
)
print(result)
```

Problem 3<br>Create a DataFrame with 'FlightID', 'Airline', 'Origin', 'Delay_Minutes', and 'Passengers', Add a 'Delay_Category' column = 'On-Time' if Delay_Minutes <= 0 else 'Delayed'., Filter to only include<br>flights with Passengers > 100, then group by Airline and Delay_Category to compute number of flights and average delay. print the final dataframe


```
data = {
    "FlightID": [1, 2, 3, 4, 5, 6],
    "Airline": ["A", "A", "B", "B", "C", "C"],
    "Origin": ["NYC", "LAX", "NYC", "LAX", "NYC", "LAX"],
    "Delay_Minutes": [5, -2, 0, 25, -5, 10],
    "Passengers": [120, 80, 150, 200, 95, 300],
}

# Solution
df = pd.DataFrame(data)
df["Delay_Category"] = df["Delay_Minutes"].apply(lambda x: "On-Time" if x <= 0 else "Delayed")
filtered_df = df[df["Passengers"] > 100]
result = filtered_df.groupby(["Airline", "Delay_Category"]).agg(
    Num_Flights=("FlightID", "count"),
    Avg_Delay=("Delay_Minutes", "mean")
)
print(result)
```

Problem 4<br>Create a DataFrame with 'Year', 'Quarter', 'Region', 'Sales', and 'Cost', Add a 'Profit' column = Sales - Cost. Filter to only include rows where Profit > 100, then group by Year, Quarter, and Region<br>to compute total profit and average sales. print the final dataframe


```
data = {
    "Year": [2023, 2023, 2023, 2024, 2024, 2024],
    "Quarter": ["Q1", "Q1", "Q2", "Q1", "Q2", "Q2"],
    "Region": ["East", "West", "East", "West", "East", "West"],
    "Sales": [500, 400, 600, 700, 300, 450],
    "Cost": [300, 350, 400, 500, 150, 400],
}

# Solution
df = pd.DataFrame(data)
df["Profit"] = df["Sales"] - df["Cost"]
filtered_df = df[df["Profit"] > 100]
result = filtered_df.groupby(["Year", "Quarter", "Region"]).agg(
    Total_Profit=("Profit", "sum"),
    Avg_Sales=("Sales", "mean")
)
print(result)
```

Problem 5<br>Create a DataFrame with 'Name' and 'Birth_Year', Write a function `calculate_age` that takes a birth year and returns age assuming the current year is 2023, Apply this function to create a new 'Age'<br>column and print the final dataframe.


```
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Birth_Year": [1990, 1985, 2000, 1975],
}
df = pd.DataFrame(data)

def calculate_age(year):
    return 2023 - year

# Solution
df["Age"] = df["Birth_Year"].apply(calculate_age)
print(df)
```

Problem 6<br>Create a DataFrame with 'Product' and 'Price', Write a function `price_category` that returns "Low" if price < 50, "Medium" if 50–100, and "High" otherwise. Apply this function to create a new<br>'Category' column and print the final dataframe.


```
import pandas as pd

data = {
    "Product": ["A", "B", "C", "D"],
    "Price": [30, 75, 120, 50],
}

# Solution
df = pd.DataFrame(data)

def price_category(price):
    if price < 50:
        return "Low"
    elif price <= 100:
        return "Medium"
    else:
        return "High"

df["Category"] = df["Price"].apply(price_category)
print(df)
```

Problem 7<br>Create a DataFrame with 'Employee' and 'Hours_Worked', Write a function `overtime` that returns "Overtime" if Hours_Worked > 40 else "Regular", Apply this function to create a new 'Status' column and<br>print the final dataframe.


```
data = {
    "Employee": ["A", "B", "C", "D"],
    "Hours_Worked": [38, 45, 40, 50],
}

# Solution
df = pd.DataFrame(data)

def overtime(hours):
    if hours > 40:
        return "Overtime"
    else:
        return "Regular"

df["Status"] = df["Hours_Worked"].apply(overtime)
print(df)
```

Problem 8<br>Create a DataFrame with 'Student' and 'Score', Write a function `grade_letter` that converts a numeric score to a letter grade, (A for >=90, B for 80-89, C for 70-79, D for 60-69, F otherwise). Apply<br>this function to create a new 'Grade' column and print the final dataframe.


```
data = {
    "Student": ["Alice", "Bob", "Charlie", "David"],
    "Score": [95, 82, 67, 55],
}

# Solution
df = pd.DataFrame(data)

def grade_letter(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"
        
df["Grade"] = df["Score"].apply(grade_letter)
print(df)
```
