Problem 10

```
data = {"Name": ["Alice", "Bob", "Charlie", "David"], "Score": [85, 92, 88, 95]}

# Solution
df = pd.DataFrame(data)


def filter_high_scores(df):
    return df[df["Score"] > 90]


high_scores_df = filter_high_scores(df)
print(high_scores_df)
```

Problem 20
Create a DataFrame with 'City' and 'Temperature' columns. Add a new column 'Temp_Celsius' that converts temperature from Fahrenheit to Celsius. Filter to show cities with temperature above 25°C.


```
data = {
    "City": ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix"],
    "Temperature": [75, 85, 70, 90, 105],
}

# Solution
df = pd.DataFrame(data)
df["Temp_Celsius"] = (df["Temperature"] - 32) * 5 / 9
filtered_df = df[df["Temp_Celsius"] > 25]
print(filtered_df)
```

Problem 21<br>Create a DataFrame with 'Name', 'Base_Salary', and 'Bonus' columns. Add a new column 'Total_Compensation' as the sum of 'Base_Salary' and 'Bonus'. Filter to show employees with 'Total_Compensation'<br>greater than the mean.


```
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Base_Salary": [50000, 60000, 55000, 65000, 58000],
    "Bonus": [5000, 7000, 6000, 8000, 6500],
}

# Solution
df = pd.DataFrame(data)
df["Total_Compensation"] = df["Base_Salary"] + df["Bonus"]
mean_compensation = df["Total_Compensation"].mean()
filtered_df = df[df["Total_Compensation"] > mean_compensation]
print(filtered_df)
```

Problem 22<br>Create a DataFrame wi th 'Country', 'GDP', and 'Population' columns. Add a new column 'GDP_per_capita' by dividing 'GDP' by 'Population'. Filter to show countries with 'GDP_per_capita' above the<br>median.


```
data = {
    "Country": ["USA", "China", "Japan", "Germany", "UK", "India"],
    "GDP": [21400000, 14700000, 5100000, 3800000, 2700000, 2900000],
    "Population": [331000000, 1439000000, 126000000, 83000000, 67000000, 1380000000],
}

# Solution
df = pd.DataFrame(data)
df["GDP_per_capita"] = df["GDP"] / df["Population"]
median_gdp_per_capita = df["GDP_per_capita"].median()
filtered_df = df[df["GDP_per_capita"] > median_gdp_per_capita]
print(filtered_df)
```
