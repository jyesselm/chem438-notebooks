Problem 1
Given a dictionary `scores` of student names and scores, print the names of students who scored above 80.


```
scores = {"Alice": 85, "Bob": 72, "Charlie": 90}

# Solution
for name, score in scores.items():
    if score > 80:
        print(name)
```

Problem 2
Given a list of numbers `nums`, write a function `square_list` and returns a list of the squares of the numbers and print it.


```
nums = [1, 2, 3, 4]

# Solution
def square_list(lst):
    new_lst = []
    for n in lst:
        new_lst.append(n ** 2)
    return new_lst

print(square_list(nums))
```

Problem 3
Given a list of numbers `numbers`, write a function `sum_above_10` that sums all numbers greater than 10 and print it.


```
numbers = [5, 15, 8, 20, 3]

# Solution
def sum_above_10(lst):
    total = 0
    for n in lst:
        if n > 10:
            total += n
    return total

print(sum_above_10(numbers))
```

Problem 4
Given a list of numbers `numbers`, write a function `categorize_numbers` that prints "small" for numbers <5, "medium" for 5-10, and "large" for >10.


```
numbers = [2, 7, 11, 5, 1]

# Solution
def categorize_numbers(lst):
    for n in lst:
        if n < 5:
            print(n, "small")
        elif n <= 10:
            print(n, "medium")
        else:
            print(n, "large")

categorize_numbers(numbers)
```

Problem 5
Given a list of words `words`, use if/elif/else to classify each word as short (<4), medium (4-6), or long (>6) and print it.


```
words = ["hi", "hello", "wonderful", "sun"]

# Solution
for w in words:
    if len(w) < 4:
        print(w, "is short")
    elif len(w) <= 6:
        print(w, "is medium")
    else:
        print(w, "is long")
```

Problem 6
Given a dictionary `inventory` of items and quantities, write a function `print_empty_items` that prints only items with quantity zero.


```
inventory = {"apple": 0, "banana": 3, "orange": 0}

# Solution
def print_empty_items(inv):
    for item, qty in inv.items():
        if qty == 0:
            print(item)

print_empty_items(inventory)
```

Problem 7
Given a list of names `names`, write a function `greet_names` that prints "Hello, name!" for each name.


```
names = ["Alice", "Bob", "Charlie"]

# Solution
def greet_names(lst):
    for name in lst:
        print(f"Hello, {name}!")

greet_names(names)
```

Problem 8
Given a dictionary `scores` of students and their scores, use a for loop to count how many students scored above 70. Store the result in `count`.


```
scores = {"Alice": 85, "Bob": 62, "Charlie": 90, "Diana": 70}

# Solution
count = 0
for student, score in scores.items():
    if score > 70:
        count += 1
print(count)
```

Problem 9
Given a list of numbers `numbers`, use a for loop to print only the numbers that are both even and greater than 10.


```
numbers = [4, 12, 15, 20, 7, 10]

# Solution
for n in numbers:
    if n % 2 == 0 and n > 10:
        print(n)
```

Problem 10
Given a dictionary `fruits` of fruit → price, use a for loop to create a list `cheap_fruits` of fruits costing less than 2. Then print the list.


```
fruits = {"apple": 2.5, "banana": 1.0, "orange": 1.5, "mango": 3.0}

# Solution
cheap_fruits = []
for fruit, price in fruits.items():
    if price < 2:
        cheap_fruits.append(fruit)
print(cheap_fruits)
```

Problem 11
Given a dictionary `grades` of students and their list of scores, write a function `average_above_80` that prints the student's name if their average score is above 80.


```
grades = {"Alice": [90, 85, 80], "Bob": [70, 60, 65], "Charlie": [88, 92, 79]}

# Solution
def average_above_80(g):
    for student, scores in g.items():
        avg = sum(scores) / len(scores)
        if avg > 80:
            print(student)

average_above_80(grades)
```

Problem 12
Given a dictionary `inventory` where each value is another dictionary with keys `price` and `quantity`,use a for loop to print items whose total value (price * quantity) exceeds 50.


```
inventory = {
    "apple": {"price": 2, "quantity": 40},
    "banana": {"price": 1, "quantity": 30},
    "mango": {"price": 5, "quantity": 5}
}

# Solution
for item, data in inventory.items():
    total_value = data["price"] * data["quantity"]
    if total_value > 50:
        print(item)
```

Problem 13
Given a list of numbers `numbers`, write a function `pair_sums_greater_than` that takes `threshold` and prints all pairs (i, j) whose sum is greater than `threshold`.


```
numbers = [3, 8, 12, 5]
threshold = 15

# Solution
def pair_sums_greater_than(lst, threshold):
    for i in lst:
        for j in lst:
            if i + j > threshold:
                print(i, j)

pair_sums_greater_than(numbers, threshold)
```

Problem 14
Given a dictionary `employees` where each key is a department and each value is a list of names, write a function `print_all_employees` that prints "department: name" for every employee.


```
employees = {
    "HR": ["Alice", "Bob"],
    "IT": ["Charlie"],
    "Finance": ["Dana", "Eli"]
}

# Solution
def print_all_employees(emp):
    for dept, names in emp.items():
        for name in names:
            print(f"{dept}: {name}")

print_all_employees(employees)
```

Problem 15
Given a dictionary `library` where keys are genres and values are lists of book titles, use nested loops to print "genre - book" for every book.


```
library = {
    "Fiction": ["Book A", "Book B"],
    "Science": ["Book C"],
    "History": ["Book D", "Book E"]
}

# Solution
for genre, books in library.items():
    for book in books:
        print(f"{genre} - {book}")
```

Problem 16
Given two dictionaries `prices` and `quantities` with the same keys, write a function `total_values` that prints each item and its total cost (price * quantity).


```
prices = {"pen": 1.5, "book": 8.0, "bag": 25.0}
quantities = {"pen": 10, "book": 3, "bag": 1}

# Solution
def total_values(pr, qty):
    for item in pr:
        total = pr[item] * qty[item]
        print(f"{item}: {total}")

total_values(prices, quantities)
```
