Problem 37<br>Define a function `is_even` that takes a number `n` as an argument. The function should return "Even" if the number is divisible by 2, otherwise return "Odd". Print out the results of calling the<br>function with 4, 5, and 6.


```
# Solution
def is_even(n):
    if n % 2 == 0:
        return "Even"
    else:
        return "Odd"


print(is_even(4))
print(is_even(5))
print(is_even(6))
```

Problem 38<br>Define a function `greet` that takes a string `time_of_day` as an argument. The function should return "Good morning" if `time_of_day` is "morning", "Good afternoon" if `time_of_day` is "afternoon",<br>and "Good evening" otherwise. Print out the results of calling the function with "morning", "afternoon", and "evening".


```
# Solution
def greet(time_of_day):
    if time_of_day == "morning":
        return "Good morning"
    elif time_of_day == "afternoon":
        return "Good afternoon"
    else:
        return "Good evening"


print(greet("morning"))
print(greet("afternoon"))
print(greet("evening"))
```

Problem 39<br>Define a function `compare_numbers` that takes two arguments `a` and `b`. The function should return "a is greater" if `a` is greater than `b`, "b is greater" if `b` is greater than `a`, or "Both are<br>equal" if they are the same. Print out the results of calling the function with 5 and 10, 10 and 5, and 5 and 5.


```
# Solution
def compare_numbers(a, b):
    if a > b:
        return "a is greater"
    elif b > a:
        return "b is greater"
    else:
        return "Both are equal"


print(compare_numbers(5, 10))
print(compare_numbers(10, 5))
print(compare_numbers(5, 5))
```
