Problem 1<br>Perform linear regression on a dataset and interpret the results. Generate a random dataset, perform linear regression using scipy.stats.linregress, plot the data points and the regression line, and<br>print the slope, intercept, and R-squared value.


```
# Solution
x = np.random.rand(50) * 10
y = 2 * x + 1 + np.random.randn(50)
slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)

plt.figure(figsize=(10, 6))
plt.scatter(x, y, label="Data points")
plt.plot(x, slope * x + intercept, color="red", label="Regression line")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression Example")
plt.legend()

print(f"Slope: {slope:.4f}")
print(f"Intercept: {intercept:.4f}")
print(f"R-squared: {r_value**2:.4f}")
```

Problem 2<br>Compare the performance of two linear regression models on different datasets. Create two datasets with different levels of noise, perform linear regression on both, and compare their R-squared<br>values. Visualize both datasets and their regression lines on the same plot with two different axes.


```
# Solution
x1 = np.random.rand(50) * 10
y1 = 3 * x1 + 2 + np.random.randn(50) * 0.5  # Low noise
x2 = np.random.rand(50) * 10
y2 = 3 * x2 + 2 + np.random.randn(50) * 2  # High noise

slope1, intercept1, r_value1, _, _ = stats.linregress(x1, y1)
slope2, intercept2, r_value2, _, _ = stats.linregress(x2, y2)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

ax1.scatter(x1, y1, label="Data points")
ax1.plot(x1, slope1 * x1 + intercept1, color="red", label="Regression line")
ax1.set_xlabel("X")
ax1.set_ylabel("Y")
ax1.set_title("Low Noise Dataset")
ax1.legend()

ax2.scatter(x2, y2, label="Data points")
ax2.plot(x2, slope2 * x2 + intercept2, color="red", label="Regression line")
ax2.set_xlabel("X")
ax2.set_ylabel("Y")
ax2.set_title("High Noise Dataset")
ax2.legend()

print("Low Noise Dataset:")
print(f"R-squared: {r_value1**2:.4f}")
print("\nHigh Noise Dataset:")
print(f"R-squared: {r_value2**2:.4f}")
```

Problem 3<br>Perform a one-sample t-test to determine if a sample of test scores is significantly different from the population mean. Generate a sample of 30 test scores with a mean of 75 and a standard deviation<br>of 8, and test if it's significantly different from the population mean of 70.


```
# Solution
sample_scores = np.random.normal(75, 8, 30)
population_mean = 70

t_statistic, p_value = stats.ttest_1samp(sample_scores, population_mean)

print("One-sample t-test results:")
print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("The sample mean is significantly different from the population mean.")
else:
    print(
        "There is no significant difference between the sample mean and the population mean."
    )
```

Problem 4<br>Perform an independent two-sample t-test to compare the means of two groups. Generate two samples of 25 values each, representing the exam scores of two different classes, and determine if there's a<br>significant difference between their means.


```
# Solution
np.random.seed(42)
class1_scores = np.random.normal(75, 10, 25)
class2_scores = np.random.normal(70, 10, 25)

t_statistic, p_value = stats.ttest_ind(class1_scores, class2_scores)

print("\nIndependent two-sample t-test results:")
print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("There is a significant difference between the means of the two classes.")
else:
    print("There is no significant difference between the means of the two classes.")
```

Problem 5
Perform a paired t-test to compare the scores of students before and after a tutoring program. Generate paired data for 20 students and determine if there's a significant improvement in their scores.


```
# Solution
np.random.seed(42)
before_scores = np.random.normal(65, 10, 20)
improvement = np.random.normal(10, 5, 20)
after_scores = before_scores + improvement

t_statistic, p_value = stats.ttest_rel(before_scores, after_scores)

print("\nPaired t-test results:")
print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("There is a significant improvement in scores after the tutoring program.")
else:
    print("There is no significant improvement in scores after the tutoring program.")

plt.figure(figsize=(10, 6))
plt.boxplot([before_scores, after_scores], labels=["Before", "After"])
plt.title("Student Scores Before and After Tutoring")
plt.ylabel("Score")
```
