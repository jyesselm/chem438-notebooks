Problem 2
Calculate the molecular weight and LogP of caffeine. Use the SMILES string "CN1C=NC2=C1C(=O)N(C(=O)N2C)C" for caffeine. Use descriptors from the rdkit.Chem.Descriptors module.


```
# Solution
caffeine_smiles = "CN1C=NC2=C1C(=O)N(C(=O)N2C)C"
mol = Chem.MolFromSmiles(caffeine_smiles)
mol_weight = Descriptors.ExactMolWt(mol)
log_p = Descriptors.MolLogP(mol)
print(f"Molecular Weight of Caffeine: {mol_weight:.2f}")
print(f"LogP of Caffeine: {log_p:.2f}")
```

Problem 3
Create a 2D depiction of nicotine and add explicit hydrogens before drawing. Use SMILES "CN1CCCC1C2=CN=CC=C2" for nicotine. Display both the H-implicit and H-explicit images side-by-side.


```
# Solution
nicotine_smiles = "CN1CCCC1C2=CN=CC=C2"
mol = Chem.MolFromSmiles(nicotine_smiles)
molH = Chem.AddHs(mol)
AllChem.Compute2DCoords(mol)
AllChem.Compute2DCoords(molH)
img1 = Draw.MolToImage(mol, size=(250, 250), legend="Nicotine (implicit H)")
img2 = Draw.MolToImage(molH, size=(250, 250), legend="Nicotine (explicit H)")
img1.show()
img2.show()
```

Problem 4<br>Count the number of hydrogen bond donors and acceptors in paracetamol (acetaminophen). Use the SMILES string "CC(=O)NC1=CC=C(C=C1)O" for paracetamol. Use descriptors from the rdkit.Chem.Descriptors<br>module.


```
# Solution
paracetamol_smiles = "CC(=O)NC1=CC=C(C=C1)O"
mol = Chem.MolFromSmiles(paracetamol_smiles)
num_hbd = Descriptors.NumHDonors(mol)
num_hba = Descriptors.NumHAcceptors(mol)
print(f"Number of H-bond donors in Paracetamol: {num_hbd}")
print(f"Number of H-bond acceptors in Paracetamol: {num_hba}")
```

Problem 5
Count aromatic atoms and aromatic bonds in naproxen. Use SMILES "CC(C)c1ccc2cc(ccc2c1)C(C)C(=O)O" for naproxen.


```
# Solution
nap_smiles = "CC(C)c1ccc2cc(ccc2c1)C(C)C(=O)O"
mol = Chem.MolFromSmiles(nap_smiles)
num_aromatic_atoms = sum(1 for a in mol.GetAtoms() if a.GetIsAromatic())
num_aromatic_bonds = sum(1 for b in mol.GetBonds() if b.GetIsAromatic())
print(f"Naproxen aromatic atoms: {num_aromatic_atoms}")
print(f"Naproxen aromatic bonds: {num_aromatic_bonds}")
```

Problem 6<br>Create a grid of 2D depictions for four common over-the-counter drugs: aspirin, ibuprofen, acetaminophen (paracetamol), and naproxen. Use their SMILES strings to create RDKit molecule objects, then<br>use RDKit's Draw.MolsToGridImage function to create a 2x2 grid of these molecules. Label each molecule with its name.


```
# Solution
aspirin_smiles = "CC(=O)OC1=CC=CC=C1C(=O)O"
ibuprofen_smiles = "CC(C)Cc1ccc(cc1)[C@H](C)C(=O)O"
acetaminophen_smiles = "CC(=O)NC1=CC=C(C=C1)O"
naproxen_smiles = "CC(C)c1ccc2cc(ccc2c1)C(C)C(=O)O"

mols = [
    Chem.MolFromSmiles(smiles)
    for smiles in [
        aspirin_smiles,
        ibuprofen_smiles,
        acetaminophen_smiles,
        naproxen_smiles,
    ]
]

names = ["Aspirin", "Ibuprofen", "Acetaminophen", "Naproxen"]
img = Draw.MolsToGridImage(mols, molsPerRow=2, subImgSize=(300, 300), legends=names)
```

Problem 7
Compute Gasteiger partial charges after adding explicit hydrogens for acetaminophen and print the charge on O and N atoms. Use SMILES "CC(=O)NC1=CC=C(C=C1)O" for acetaminophen.


```
# Solution
apap = Chem.MolFromSmiles("CC(=O)NC1=CC=C(C=C1)O")
apapH = Chem.AddHs(apap)
AllChem.ComputeGasteigerCharges(apapH)
print("Acetaminophen partial charges (O and N atoms):")
for atom in apapH.GetAtoms():
    if atom.GetSymbol() in {"O", "N"}:
        q = float(atom.GetProp("_GasteigerCharge"))
        print(f"  Atom {atom.GetIdx()} {atom.GetSymbol()}: {q:.3f}")
```

Problem 8
Count ring sizes present in morphine by listing unique ring sizes found in the SSSR. Use SMILES "CN1CCC23C4C1CC5=CC(=O)OC(C=C5C2=CC=C3O)O" for morphine.


```
# Solution
morphine_smiles = "CN1CCC23C4C1CC5=CC(=O)OC(C=C5C2=CC=C3O)O"
mol = Chem.MolFromSmiles(morphine_smiles)
ri = mol.GetRingInfo()
atom_rings = ri.AtomRings()
ring_sizes = sorted({len(r) for r in atom_rings})
print("Morphine ring sizes in SSSR:", ring_sizes)
```

Problem 9
Calculate the number of carbon and oxygen atoms in the aspirin molecule. Use a loop to iterate over the atoms in the molecule and count the occurrences of carbon and oxygen.


```
# Solution
aspirin_smiles = "CC(=O)OC1=CC=CC=C1C(=O)O"
mol = Chem.MolFromSmiles(aspirin_smiles)

carbon_count = 0
oxygen_count = 0

for atom in mol.GetAtoms():
    if atom.GetSymbol() == "C":
        carbon_count += 1
    elif atom.GetSymbol() == "O":
        oxygen_count += 1

print(f"Number of carbon atoms in aspirin: {carbon_count}")
print(f"Number of oxygen atoms in aspirin: {oxygen_count}")
```

Problem 10
Find all atoms in the caffeine molecule that are part of a ring system. Use a loop to iterate over the atoms and check if they are in a ring. Print the atomic number and symbol for each ring atom.


```
# Solution
caffeine_smiles = "CN1C=NC2=C1C(=O)N(C(=O)N2C)C"
mol = Chem.MolFromSmiles(caffeine_smiles)

print("Atoms in ring systems of caffeine:")
for atom in mol.GetAtoms():
    if atom.IsInRing():
        print(f"Atomic number: {atom.GetAtomicNum()}, Symbol: {atom.GetSymbol()}")
```

Problem 11
Highlight specific atoms in a molecule using RDKit. Create a molecule of benzene, then highlight three alternating carbon atoms in red.


```
# Solution
benzene_smiles = "c1ccccc1"
mol = Chem.MolFromSmiles(benzene_smiles)
highlight_atoms = [0, 2, 4]  # Indices of atoms to highlight
highlight_colors = [(1, 0, 0)] * len(highlight_atoms)

img = Draw.MolToImage(
    mol,
    size=(300, 300),
    highlightAtoms=highlight_atoms,
    highlightColor=highlight_colors,
)
```

Problem 12
Set custom atom labels for a molecule using RDKit. Create a molecule of ethanol and label the oxygen atom as "OH" and the carbon atoms as "C1" and "C2".


```
# Solution
ethanol_smiles = "CCO"
mol = Chem.MolFromSmiles(ethanol_smiles)

atom_labels = {0: "C1", 1: "C2", 2: "OH"}

for atom in mol.GetAtoms():
    if atom.GetIdx() in atom_labels:
        atom.SetProp("atomLabel", atom_labels[atom.GetIdx()])

img = Draw.MolToImage(mol, size=(300, 300))
img.show()
```

Problem 13
Create a molecule of benzene and label every other carbon atom with its atomic number. Display the resulting molecule.


```
# Solution
benzene_smiles = "c1ccccc1"
mol = Chem.MolFromSmiles(benzene_smiles)

for atom in mol.GetAtoms():
    if atom.GetIdx() % 2 == 0:
        atom.SetProp("atomLabel", str(atom.GetAtomicNum()))

img = Draw.MolToImage(mol, size=(300, 300))
img.show()
```

Problem 14
Perform molecular fragmentation using RDKit. Create a molecule of aspirin, break the ester bond, and display the resulting fragments.


```
# Solution
aspirin_smiles = "CC(=O)OC1=CC=CC=C1C(=O)O"
mol = Chem.MolFromSmiles(aspirin_smiles)
bond_to_break = mol.GetBondBetweenAtoms(3, 4)  # Indices may vary, adjust if needed
fragments = Chem.FragmentOnBonds(mol, [bond_to_break.GetIdx()])
AllChem.Compute2DCoords(fragments)
img = Draw.MolToImage(mol, size=(300, 300), legend="Aspirin")
img_fragments = Draw.MolToImage(fragments, size=(300, 300), legend="Fragments")
```
