# Solutions — Nested loops & pairwise iteration

_Worked answers for the CHEM 438 practice set. Try each problem yourself first — you learn far more from a wrong attempt than from reading the answer._


Problem 1
Given a list of numbers, use a nested for loop to print all pairs of numbers.


```
numbers = [1, 2, 3]

# Solution
for i in numbers:
    for j in numbers:
        print(i, j)
```

Problem 2
Given a list of names, use a nested for loop to print every possible pair of different names.


```
names = ["Alice", "Bob", "Charlie"]

# Solution
for n1 in names:
    for n2 in names:
        if n1 != n2:
            print(n1, "-", n2)
```

Problem 3
Given a list of numbers, use a nested for loop to print the sum of every pair.


```
nums = [2, 4, 6]

# Solution
for i in nums:
    for j in nums:
        print(f"{i} + {j} = {i + j}")
```

Problem 4
Given a list of words, use a nested for loop to print all pairs where the first word is shorter than the second.


```
words = ["hi", "hello", "python"]

# Solution
for w1 in words:
    for w2 in words:
        if len(w1) < len(w2):
            print(w1, "is shorter than", w2)
```

Problem 5
Given a list of cities, use a nested for loop to print all pairs that start with the same letter.


```
cities = ["Paris", "Prague", "London", "Lisbon"]

# Solution
for c1 in cities:
    for c2 in cities:
        if c1 != c2 and c1[0] == c2[0]:
            print(c1, "and", c2, "start with the same letter")
```

Problem 6
Use a nested for loop to compare every pair of students and print who has the higher grade.


```
grades = {"Alice": 85, "Bob": 92, "Charlie": 78}

# Solution
for name1, grade1 in grades.items():
    for name2, grade2 in grades.items():
        if name1 != name2 and grade1 > grade2:
            print(f"{name1} has a higher grade than {name2}")
```

Problem 7
Use a nested for loop to find and print any two cities that have the same population.


```
cities = {"Omaha": 500000, "Austin": 950000, "Denver": 500000}

# Solution
for city1, pop1 in cities.items():
    for city2, pop2 in cities.items():
        if city1 != city2 and pop1 == pop2:
            print(f"{city1} and {city2} have the same population ({pop1})")
```

Problem 8
Use a nested for loop to print all possible item pairs and their combined cost.


```
prices = {"apple": 1.0, "banana": 0.5, "pear": 2.0}

# Solution
for item1, price1 in prices.items():
    for item2, price2 in prices.items():
        if item1 != item2:
            print(f"{item1} + {item2} = ${price1 + price2}")
```

Problem 9
Given a list of numbers, use a nested for loop to print only pairs whose sum is even.


```
numbers = [1, 2, 3, 4]

# Solution
for i in numbers:
    for j in numbers:
        if (i + j) % 2 == 0:
            print(f"{i} + {j} = {i + j} (even)")
```

Problem 10
Use a nested for loop to print all item pairs and their combined cost if the total is less than $10.


```
prices = {"apple": 2, "banana": 3, "pear": 5, "grape": 8}

# Solution
for item1, p1 in prices.items():
    for item2, p2 in prices.items():
        if item1 != item2 and (p1 + p2) < 10:
            print(f"{item1} + {item2} = ${p1 + p2}")
```

Problem 11
Use a nested for loop to print which students pass each threshold.


```
grades = {"Alice": 85, "Bob": 72, "Charlie": 90}
thresholds = [60, 80, 90]

# Solution
for student, grade in grades.items():
    for t in thresholds:
        if grade >= t:
            print(f"{student} passes threshold {t}")
```

Problem 12
Use a nested for loop to calculate the total cost of the shopping list.


```
prices = {"apple": 2, "banana": 3, "pear": 5}
shopping_list = ["apple", "pear", "banana", "apple"]

# Solution
total = 0
for item in shopping_list:
    for name, price in prices.items():
        if item == name:
            total += price
print(f"Total cost: ${total}")
```

Problem 13
Use a nested for loop to add 5 points to those students and print the new score.


```
scores = {"Alice": 80, "Bob": 85, "Charlie": 70}
bonus_students = ["Bob", "Charlie"]

# Solution
for name in bonus_students:
    for student, score in scores.items():
        if name == student:
            print(f"{student}'s new score is {score + 5}")
```

Problem 14
Use a nested for loop to print each student's email by matching across both dicts.


```
student_id = {"Alice": "S1", "Bob": "S2", "Charlie": "S3"}
id_email = {
    "S1": "alice@school.edu",
    "S2": "bob@school.edu",
    "S3": "charlie@school.edu",
}

# Solution
for name, sid in student_id.items():
    for iid, email in id_email.items():
        if sid == iid:
            print(f"{name}: {email}")
```

Problem 15
Use a nested for loop to print all city/letter pairs that match.


```
cities = ["Paris", "Prague", "London", "Lisbon", "Tokyo"]
starts_with = ["P", "L", "S"]

# Solution
for letter in starts_with:
    for city in cities:
        if city[0] == letter:
            print(f"{city} starts with {letter}")
```

Problem 16
Use a for loop with f-strings to print: "Alice scored 85 points".


```
grades = {"Alice": 85, "Bob": 72, "Charlie": 90}

# Solution
for student, score in grades.items():
    print(f"{student} scored {score} points")
```

Problem 17
Use a for loop with f-strings to print: "The price of apple is $2".


```
prices = {"apple": 2, "banana": 3, "pear": 5}

# Solution
for item, cost in prices.items():
    print(f"The price of {item} is ${cost}")
```
